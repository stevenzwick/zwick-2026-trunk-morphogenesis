```python
rows = []

# FOXF1
for v in p3_FOXF1_score:
    rows.append(("unilateral bead", "FOXF1", v))
for v in p1_FOXF1_score:
    rows.append(("medial bead", "FOXF1", v))
for v in bibead_FOXF1_score:
    rows.append(("bilateral bead", "FOXF1", v))

# PAX8
for v in p3_PAX8_score:
    rows.append(("unilateral bead", "PAX8", v))
for v in p1_PAX8_score:
    rows.append(("medial bead", "PAX8", v))
for v in bibead_PAX8_score:
    rows.append(("bilateral bead", "PAX8", v))

# MESP2
for v in p3_MESP2_score:
    rows.append(("unilateral bead", "MESP2", v))
for v in p1_MESP2_score:
    rows.append(("medial bead", "MESP2", v))
for v in bibead_MESP2_score:
    rows.append(("bilateral bead", "MESP2", v))

data = pd.DataFrame(rows, columns=["Group", "Channel", "Score"])

# 🎨 your color scheme
# palette = {
#     "FOXF1": "#00FFFF",   # cyan
#     "PAX8": "#FFD700",    # yellow (gold, looks better than pure yellow)
#     "MESP2": "#FF0000"    # red
# }

plt.figure(figsize=(9, 5))

ax = sns.boxplot(
    data=data,
    x="Group",
    y="Score",
    hue="Channel",
    # palette = palette,
    palette=["blue", "green", "red"],
    width=0.5,
    boxprops=dict(alpha=0.3),
    fliersize=0,
    linewidth=1.5,
)

sns.stripplot(
    data=data,
    x="Group",
    y="Score",
    hue="Channel",
    # palette = palette,
    palette=["blue", "green", "red"],
    dodge=True,
    jitter=0.18,
    size=4,
    edgecolor="black",
    linewidth=0.5,
    alpha=0.3,
)

# remove duplicate legend
handles, labels = ax.get_legend_handles_labels()
ax.legend(
    handles[:3], labels[:3], title="Channel", frameon=False, bbox_to_anchor=(1.02, 1)
)

plt.ylabel("bilaterality_index_mean")
plt.xlabel("")
plt.ylim(0, 1)

plt.tight_layout()
plt.savefig("Orgs_bilaterality_score_three_channels.pdf", format="pdf", dpi=300)
plt.show()
```
